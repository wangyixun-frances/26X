# ReadMe

Loads all the sources listed under China from the Factiva website (accessed Aug 10, 2026), and filters down to mainland china media sources to see how many there are.

```python
import pandas as pd
import os

from IPython.core.interactiveshell import InteractiveShell
InteractiveShell.ast_node_interactivity = "all"
```

```python
with open("../data/factiva_china_sources.csv", "r") as raw_text:
    treated_text = raw_text.read().splitlines()  # reads each line as one string

df = pd.DataFrame(treated_text, columns=["media"])
```

```python
df.shape[0]
df.head()
df.iloc[0]
```

```python
# filters to sources only in Simplified Chinese, meaning it's from mainland China.
df_filtered = df[ df.media.str.contains("Simplified") ]
```

```python
df_filtered.shape[0]
df_filtered.sample(10)

# check if it contains People's Daily (one of the most importnat newspapers)
df_filtered[ df_filtered.media.str.contains("Daily") ]
```

# Conclusion
Factiva offers 425 different sources on chinese mainland media.
